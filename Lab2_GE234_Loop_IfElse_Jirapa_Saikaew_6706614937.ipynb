## 🔹 ตัวอย่างที่ 1: การใช้ `for loop` เพื่อแสดงค่าพิกัด


```python

# พิกัดของสถานที่สำคัญในประเทศไทย
locations = {
    "Bangkok": (13.7563, 100.5018),
    "Chiang Mai": (18.7883, 98.9853),
    "Phuket": (7.8804, 98.3923),
    "Ubon Ratchathani": (15.2287, 104.8564)
}

# วนลูปแสดงค่าพิกัด
for city, coord in locations.items():
    print(f"{city}: Latitude {coord[0]}, Longitude {coord[1]}")
```


## 🔹 ตัวอย่างที่ 2: การใช้ `if-else` กรองข้อมูลพิกัด


```python

for city, coord in locations.items():
    if coord[0] > 16:
        print(f"{city} อยู่ทางภาคเหนือของประเทศไทย")
    else:
        print(f"{city} ไม่ใช่จังหวัดทางภาคเหนือ")
```


## 🔹 ตัวอย่างที่ 3: ใช้ `while loop` นับจำนวนจุดข้อมูล


```python

points = [(13.7563, 100.5018), (18.7883, 98.9853), (7.8804, 98.3923)]
count = 0

while count < len(points):
    print(f"Point {count + 1}: {points[count]}")
    count += 1
```


## 🔹 ตัวอย่างที่ 4: วิเคราะห์ข้อมูล Raster ด้วย `for loop`


```python

import numpy as np

# จำลองข้อมูล NDVI (Normalized Difference Vegetation Index)
ndvi = np.array([
    [0.2, 0.4, 0.5],
    [0.1, 0.3, 0.7],
    [0.6, 0.8, 0.2]
])

# วนลูปตรวจสอบค่าพืชพรรณ
for row in range(ndvi.shape[0]):
    for col in range(ndvi.shape[1]):
        if ndvi[row, col] > 0.3:
            print(f"พิกัด ({row}, {col}) มีพืชพรรณหนาแน่น (NDVI={ndvi[row, col]})")
```


# 📝 **กิจกรรมในแลป**

3. **แบบฝึกหัด 3**: ใช้ `while loop` เพื่อวนลูปและแสดงค่าละติจูดจากลิสต์จนกว่าจะพบค่าต่ำกว่า 10°
4. **แบบฝึกหัด 4**: ใช้ `for loop` วิเคราะห์ค่า NDVI และหาค่าพื้นที่ที่มีสภาพแห้งแล้ง (NDVI < 0.2)


### **แบบฝึกหัด 1: แสดงชื่อจังหวัดและพิกัดจากลิสต์ของคุณเอง**

```python

my_locations = {
    "Nakhon Ratchasima": (14.9750, 102.0987),
    "Khon Kaen": (16.4444, 102.8350),
    "Chon Buri": (13.3644, 100.9840),
    "Surat Thani": (9.1398, 99.3283)
}


print("--- จังหวัดและพิกัดจากลิสต์ของฉัน ---")
for city, coord in my_locations.items():
    print(f"{city}: Latitude {coord[0]}, Longitude {coord[1]}")
```

### **แบบฝึกหัด 2: ใช้ `if-else` เพื่อแยกแยะพื้นที่ที่เป็นภูเขาและพื้นที่ราบจากลิสต์ของพิกัด**

```python

geographic_points = [
    ("Doi Inthanon", 18.5881, 98.4841),
    ("Phu Kradueng", 16.8833, 101.8500),
    ("Ayutthaya", 14.3500, 100.5667),
    ("Trang", 7.5500, 99.6167),
    ("Chiang Rai", 19.9080, 99.8333)
]

print("--- การแยกแยะพื้นที่ภูเขาและที่ราบ ---")
for name, lat, lon in geographic_points:
    if lat > 15:
        print(f"{name} (Latitude: {lat}) เป็นพื้นที่ภูเขา")
    else:
        print(f"{name} (Latitude: {lat}) เป็นพื้นที่ราบ")
```

### **แบบฝึกหัด 3: ใช้ `while loop` เพื่อวนลูปและแสดงค่าละติจูดจากลิสต์จนกว่าจะพบค่าต่ำกว่า 10°**

```python

coordinates_list = [
    ("Chiang Mai", 18.7883, 98.9853),
    ("Bangkok", 13.7563, 100.5018),
    ("Surat Thani", 9.1398, 99.3283),
    ("Phuket", 7.8804, 98.3923),
    ("Ayutthaya", 14.3500, 100.5667)
]

print("--- แสดงค่าละติจูดจนกว่าจะพบค่าต่ำกว่า 10° ---")

index = 0

while index < len(coordinates_list):
    name, lat, lon = coordinates_list[index]

    if lat < 10:
        print(f"พบค่าละติจูดที่น้อยกว่า 10° ที่ {name}: Latitude {lat} หยุดการวนลูป")
        break
    else:
        print(f"{name}: Latitude {lat}")

    index += 1


if index == len(coordinates_list) and lat >= 10:
    print("วนลูปจนครบทุกพิกัดแล้ว และไม่พบค่าละติจูดที่ต่ำกว่า 10° ในลิสต์นี้")
```

### **แบบฝึกหัด 4: ใช้ `for loop` วิเคราะห์ค่า NDVI และหาค่าพื้นที่ที่มีสภาพแห้งแล้ง (NDVI < 0.2)**

```python
import numpy as np

ndvi_data = np.array([
    [0.25, 0.45, 0.15], # Row 0
    [0.10, 0.30, 0.70], # Row 1
    [0.60, 0.05, 0.20]  # Row 2
])

print("--- การวิเคราะห์พื้นที่แห้งแล้งจากค่า NDVI ---")


for row in range(ndvi_data.shape[0]):
    for col in range(ndvi_data.shape[1]):

        if ndvi_data[row, col] < 0.2:
            print(f"พิกัด ({row}, {col}) มีสภาพแห้งแล้ง (NDVI={ndvi_data[row, col]})")
```
